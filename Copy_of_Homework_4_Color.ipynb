## Categorical Colors

### Plot 1

Error:

in this graph color is being used to encode Major_Genre, but there are too many genres for color to be effective. The color palette creates many similar colors that are difficult to distinguish, especially with heavy overplotting. This makes it hard to identify patterns or compare genres. Since position is a much more accurate encoding than color, the large number of colors distracts afrom the actual relationship between IMDB votes and ratings. Color here adds clutter rather than clarity.

Improvement:
Reduce reliance on color by limiting the number of genres shown, using opacity, or faceting by genre.

```python


p1orig = alt.Chart(movies).mark_circle().encode(
    alt.X('IMDB_Votes:Q', scale = alt.Scale(type='log')),
    alt.Y('IMDB_Rating:Q'),
    alt.Color('Major_Genre:N', scale = alt.Scale(scheme='tableau20'))
).properties(title='Original')
#
p1updated = alt.Chart(movies).mark_circle().encode(
    alt.X('IMDB_Votes:Q', scale = alt.Scale(type='log')),
    alt.Y('IMDB_Rating:Q'),
).properties(title='Updated')
#
(p1orig | p1updated).resolve_scale(color='independent')

```

### Plot 2

Error:
This plot uses a rainbow color scale to encode the quantitative variable Cylinders. Rainbow color maps are not perceptually ordered, so similar values do not look similar, and viewers may perceive artificial breaks in the data. This makes it difficult to accurately compare values across lines.

Improvement:
Use a sequential color scale that varies smoothly in lightness, or treat Cylinders as a categorical variable and use a small, discrete color palette. Line style or faceting would also be more effective than color.

```python
p2orig = alt.Chart(cars).mark_line().encode(
    alt.X('year(Year):T'),
    alt.Y('mean(Acceleration):Q'),
    alt.Color('Cylinders:Q', scale = alt.Scale(scheme='rainbow'))
).properties(title='Original')
#
p2updated = alt.Chart(cars).mark_line().encode( alt.X('year(Year):T'),
            alt.Y('mean(Acceleration):Q'),
            alt.Color('Cylinders:N') ).properties(title='Updated (Sequential Color)')
#
(p2orig | p2updated).resolve_scale(color='independent')
```

### Plot 3

Error:

The choropleth map uses multiple unrelated colors to represent binned quantitative values. The colors do not follow a clear perceptual order, and some colors (like red) visually stand out more than others regardless of value. This makes it difficult to accurately interpret which states have higher or lower values.

Improvement:
Use a single-hue sequential color scale where lightness increases with value. This would make it easier to compare magnitudes across states and better match how people perceive quantitative color.

```python
p3orig = alt.Chart(states).mark_geoshape(stroke='white').encode(
    alt.Color('engineers:Q', bin=True,
              scale=alt.Scale(domain=[0,0.002,0.004,0.006,0.008,0.01],
                              range=['#C5C5C5','#378FA3','#C60000','#FEC600','#007AC5']))
).transform_lookup(
    lookup='id',
    from_=alt.LookupData(stateinfo, 'id', ['engineers'])
).properties(width=500, height=300, title='Original').project('albersUsa')
#
p3updated = p3updated = (
    alt.Chart(states)
    .mark_geoshape(stroke='white')
    .encode(
        alt.Color(
            'engineers:Q',
            scale=alt.Scale(
                scheme='blues',        # single-hue sequential scale
                domain=[0, 0.01]       # fixed domain for consistency
            ),
            title='Proportion of Engineers'
        )
    )
    .transform_lookup(
        lookup='id',
        from_=alt.LookupData(stateinfo, 'id', ['engineers'])
    )
    .properties(
        width=500,
        height=300,
        title='Updated (Single-Hue Sequential Scale)'
    )
    .project('albersUsa')
)

#
(p3orig | p3updated).resolve_scale(color='independent')
```

### Plot 4

Error:

Each stock symbol is assigned a different color, resulting in too many lines and too many colors on the same chart. Many of the colors are difficult to distinguish, and the visual clutter makes it hard to follow any individual stock over time. All symbols are treated as equally important, which overwhelms the viewer.

Improvement:
Highlight one or two stocks using color and mute the rest in gray, or facet by symbol. This reduces clutter and establishes a clearer visual hierarchy.

```python
p4orig = alt.Chart(stocks).mark_line().encode(
    alt.X('date:T'),
    alt.Y('price:Q'),
    alt.Color('symbol:N')
).properties(width=600, title='Original')

#
p4updated = alt.Chart(stocks).transform_filter(
    alt.FieldEqualPredicate(field='symbol', equal='AAPL')
).mark_line(color='steelblue', size=3).encode(
    alt.X('date:T'),
    alt.Y('price:Q')

).properties(width=600, title='Updated (Only AAPL)')
#

(p4orig | p4updated).resolve_scale(color='independent')
```

### Plot 5

Error:

Color is used to encode MPAARating, but the same variable is already encoded by faceting rows. This makes the color redundant and unnecessary. The strong colors draw attention away from the actual comparisons, which are better communicated by bar length.

Improvement:
Remove the color encoding and use a single neutral color for all bars. Faceting alone is sufficient to distinguish MPAA ratings.

```python
p5orig = alt.Chart(movies).transform_filter(
    'datum.IMDB_Rating != null'
).transform_filter(
    alt.FieldOneOfPredicate(field='MPAA_Rating', oneOf=['G','PG','PG-13','R'])
).transform_filter(
    alt.FieldOneOfPredicate(field='Major_Genre', oneOf=['Action','Comedy','Drama','Horror'])
).mark_bar().encode(
    alt.Y('Major_Genre:N'),
    alt.X('mean(IMDB_Rating):Q'),
    alt.Row('MPAA_Rating:N'),
    alt.Color('MPAA_Rating:N')
).properties(title='Original')
#
p5updated = alt.Chart(movies).transform_filter(
    'datum.IMDB_Rating != null'
).transform_filter(
    alt.FieldOneOfPredicate(field='MPAA_Rating', oneOf=['G','PG','PG-13','R'])
).transform_filter(
    alt.FieldOneOfPredicate(field='Major_Genre', oneOf=['Action','Comedy','Drama','Horror'])
).mark_bar(color='steelblue').encode(
    alt.Y('Major_Genre:N'),
    alt.X('mean(IMDB_Rating):Q'),
    alt.Row('MPAA_Rating:N'),
    alt.Color('Major_Genre:N')
).properties(title='Updated (Removed Redundant Color)')
#
(p5orig | p5updated).resolve_scale(color='independent')
```

### Plot 6

Error:

The plot assigns a different color to each state using a sequential color scale, even though state is a categorical variable with no inherent order. This implies a false quantitative meaning and adds unnecessary visual noise.

Improvement:
Use a single color for all bars, or only highlight specific states if needed. Bar length alone is enough to communicate population differences.

```python
p6orig = alt.Chart(stateinfo).mark_bar().encode(
    alt.Y('state:N'),
    alt.X('population:Q'),
    alt.Color('state:O', scale=alt.Scale(scheme='plasma'))
).properties(title='Original')
#
p6updated = alt.Chart(stateinfo).mark_bar(color='steelblue').encode(
    alt.Y('state:N', sort ='-x'),
    alt.X('population:Q')
).properties(title='Updated (Single Color)')
#

(p6orig | p6updated).resolve_scale(color='independent')
```

## Quantitative Colors

### Plot 7

Error:

Acceleration is encoded using color even though the bars are already sorted by acceleration. This creates a redundant encoding that does not add a  new information. Since position and ordering are more accurate encodings than color, the color channel is not really needed here.

Improvement:

 I removed the color encoding entirely and used a single neutral color for all bars. This reduces cognitive load and keeps the viewers attention on the meaningful encoding: bar length. Since the bars are already sorted by Acceleration, position communicates the ranking more effectively than color.

```python
# Original
p7orig = (
    alt.Chart(cars)
    .mark_bar()
    .transform_filter('datum.Cylinders == 8 & year(datum.Year) == 1975')
    .encode(
        alt.X('Weight_in_lbs:Q', title='Weight (lbs)'),
        alt.Y('Name:N', sort=alt.Sort(field='Acceleration')),
        alt.Color('Acceleration:Q')
    )
    .properties(title='Original')
)

# Updated 1: Name vs Weight (removed redundant color)
p7updated_weight = (
    alt.Chart(cars)
    .mark_bar(color='steelblue')
    .transform_filter('datum.Cylinders == 8 & year(datum.Year) == 1975')
    .encode(
        alt.X('Weight_in_lbs:Q', title='Weight (lbs)'),
        alt.Y('Name:N', sort=alt.Sort(field='Weight_in_lbs', order='descending'))
    )
    .properties(title='Updated: Name vs Weight')
)

# Updated 2: Name vs Acceleration
p7updated_accel = (
    alt.Chart(cars)
    .mark_bar(color='steelblue')
    .transform_filter('datum.Cylinders == 8 & year(datum.Year) == 1975')
    .encode(
        alt.X('Acceleration:Q', title='Acceleration'),
        alt.Y('Name:N', sort=alt.Sort(field='Acceleration', order='descending'))
    )
    .properties(title='Updated: Name vs Acceleration')
)

# Combine side by side
(p7orig | p7updated_weight | p7updated_accel)
```

### Plot 8

Error:

The heatmap uses a redgreen color scheme to encode quantitative values, which is not accessible for viewers with color vision deficiencies. Additionally, the color scale does not change evenly in lightness, making some differences appear more important than others.

Improvement:

I replaced the original palette with the viridis sequential color scheme. Viridis is perceptually uniform, colorblindsafe, and increases smoothly in lightness, making it much easier to compare MPG values across the heatmap. This improves both accessibility and interpretability without changing the underlying data or structure of the plot.

```python
p8orig = alt.Chart(cars).transform_joinaggregate(
    groupby=['Name'],
    yearcount='count()'
).transform_filter(
    'datum.yearcount > 2'
).mark_rect().encode(
    alt.X('year(Year):T', title='Year'),
    alt.Y('Name:N'),
    alt.Color('max(Miles_per_Gallon):Q',
              scale=alt.Scale(domain=[10,25,40],
                              range=['#FB261B','#F8F1CE','#1D79E0']))
).properties(title='Original')

p8updated = alt.Chart(cars).transform_joinaggregate(
    groupby=['Name'],
    yearcount='count()'
).transform_filter(
    'datum.yearcount > 2'
).mark_rect().encode(
    alt.X('year(Year):T', title='Year'),
    alt.Y('Name:N'),
    alt.Color('max(Miles_per_Gallon):Q',
              scale=alt.Scale(scheme='viridis'))
).properties(title='Updated (Viridis)')

p8orig | p8updated
```

### Plot 9

Error:

Miles per gallon is encoded using color, but the color scale does not provide a clear perceptual ordering. Color dominates the visualization even though the main comparisons are better supported by positionthe color encoding makes the plot harder to interpret.

Improvement:


We should use a more subtle sequential color scale or remove color entirely and rely on position or faceting to show differences in MPG.

```python
p9orig = alt.Chart(cars).mark_circle().encode(
    alt.X('Horsepower:Q'),
    alt.Y('Origin:N'),
    alt.Color('Miles_per_Gallon:Q')
).transform_calculate(
    jitter="sqrt(-2*log(random()))*cos(2*PI*random())"
).properties(
    height=300,
    title='Original'
)

# Updated
p9updated = alt.Chart(cars).mark_circle(size=70, opacity=0.6, color='steelblue').encode(
    alt.X('Horsepower:Q'),
    alt.Y('Origin:N')
).transform_calculate(
    jitter="sqrt(-2*log(random()))*cos(2*PI*random())"
).properties(
    height=300,
    title='Updated (Removed Color, Rely on Position)'
)
#

p9orig | p9updated
```

Error:

In the original plot, I used color to encode State, even though the comparison only involves a small set of states. While four categories are easier to distinguish than dozens, color still becomes visually dominant and distracts from the actual relationship between applications and tuition. The plot relies too heavily on hue differences, even though position already communicates the structure more effectively.

Improvement:  


I removed the color encoding entirely and used a single neutral color. This reduces visual noise and shifts attention back to the positional relationship between the number of applications and instate tuition. Relying on position instead of color makes the visualization cleaner, more readable, and more aligned with perceptual best practices.

```python

import altair as alt
import pandas as pd
dataurl = 'http://lib.stat.cmu.edu/datasets/colleges/usnews.data'
college = pd.read_csv(
    dataurl,
    na_values='*',
    names=[
        'FICE','College','State','Private','MathSAT','VerbalSAT','CombinedSAT','ACT',
        'Q1Math','Q3Math','Q1Verbal','Q3Verbal','Q1ACT','Q3ACT',
        'Applied','Accepted','Enrolled','Top10HS','Top25HS','FullTimeUG','PartTimeUG',
        'InStateTuition','OutOfStateTuition','RoomBoardCost','RoomCost','BoardCost',
        'ExtraFees','BookFees','PersonalFees',
        'PhDFaculty','TerminalFaculty','StuFacRatio','AlumniDonate','SpendPerStudent','GradRate'
    ]
)

states_of_interest = ["WA", "OR", "ID", "AK"]
college_subset = college[college["State"].isin(states_of_interest)]

p10orig = alt.Chart(college_subset).mark_circle().encode(
    alt.X('Applied:Q', title='Number of Applications'),
    alt.Y('InStateTuition:Q', title='In-State Tuition, USD'),
    alt.Color('State:N')
).properties(
    width=500,
    height=350,
    title='Original: Applications vs Tuition (Colored by State)'
)

#
p10updated = alt.Chart(college_subset).mark_circle(
    size=70,
    opacity=0.6,
    color='steelblue'
).encode(
    alt.X('Applied:Q', title='Number of Applications'),
    alt.Y('InStateTuition:Q', title='In-State Tuition, USD')
).properties(
    width=500,
    height=350,
    title='Updated: Removed Color, Rely on Position'
)
#

p10orig | p10updated
```
